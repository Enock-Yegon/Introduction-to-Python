# Module 1: Pandas Foundations & Data Inspection

## 1. Core Concepts
* **Series**: A single column of data (1-Dimensional array).
* **DataFrame**: A two-dimensional table made of rows and columns (like an Excel sheet or SQL table).
  * **Rows**: Identified by index numbers starting at $0$.
  * **Columns**: Identified by header names (e.g., `"Name"`, `"Score"`).
  

```python
import pandas as pd

# 1. Raw data represented as a dictionary
class_data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Enock"],
    "Score": [88, 72, 95, 60, 84],
    "Passed": [True, True, True, False, True],
}

# 2. Convert dictionary into a Pandas DataFrame
df = pd.DataFrame(class_data)

# 3. Display the DataFrame
df
```

## 2. Summary of Essential Inspection Tools

| Command | Purpose | Output / Example |
| :--- | :--- | :--- |
| `pd.DataFrame(data)` | Converts raw data (e.g., a dictionary) into a table. | `df = pd.DataFrame(data)` |
| `df.head(n)` | View the first $n$ rows (default is 5). | `df.head(3)` |
| `df.tail(n)` | View the last $n$ rows. | `df.tail(3)` |
| `df.shape` | Returns dimensions as a tuple `(rows, columns)`. | `(5, 4)` |
| `df.info()` | Summary of column names, data types (`dtypes`), and non-null counts. | Technical details |
| `df.describe()` | Key summary statistics ($\mu, \sigma$, min, max, quartiles) for numeric columns. | Statistical table |
| `len(df)` | Quick count of total rows. | Integer count |

```python
# View the first 3 rows
print("--- HEAD ---")
print(df.head(3))

# Check dimensions (rows, columns)
print("\n--- SHAPE ---")
print(df.shape)  # Returns a tuple: (5, 4)

# Get column names and data types
print("\n--- INFO ---")
df.info()

# Summary statistics for numerical columns
print("\n--- DESCRIBE ---")
print(df.describe())
```

# Module 3: Data Cleaning & Handling Missing Values (NaN)

## 1. Key Commands for Missing Data

| Command | Action / Purpose |
| :--- | :--- |
| `df.isnull()` or `df.isna()` | Returns `True` for every cell that contains missing/NaN data. |
| `df.isnull().sum()` | Counts total missing values in **each column**. |
| `df.dropna()` | **Removes/drops** any row that contains at least one `NaN`. |
| `df.fillna(value)` | **Replaces/fills** missing values with a default value (e.g., mean or `0`). |
| `df.drop(columns=["ColName"])` | Deletes an entire unwanted column from the DataFrame. |

> **Crucial Concept (`inplace=True`):**
> Most Pandas functions **do not modify** the original DataFrame by default—they return a copy. To modify the DataFrame permanently, add `inplace=True` (e.g., `df.fillna(0, inplace=True)`).

```python
import numpy as np
import pandas as pd

# Sample messy dataset with missing values (NaN)
raw_data = {
    "Laptop": ["MacBook", "Dell XPS", "ThinkPad", "HP Spectre", "Asus ZenBook"],
    "Price_USD": [1200, np.nan, 950, 1100, np.nan],
    "RAM_GB": [16, 8, np.nan, 16, 8],
    "Storage_GB": [512, 256, 512, np.nan, 256],
}

df = pd.DataFrame(raw_data)

print("=== 1. RAW DATASET WITH MISSING VALUES ===")
print(df)

print("\n=== 2. COUNT MISSING VALUES PER COLUMN ===")
print(df.isnull().sum())

# Strategy A: Fill missing prices with the average (mean) price
mean_price = df["Price_USD"].mean()
df["Price_USD"] = df["Price_USD"].fillna(mean_price)

# Strategy B: Fill missing RAM with a fixed default value (8 GB)
df["RAM_GB"] = df["RAM_GB"].fillna(8)

# Strategy C: Drop rows where critical info (Storage_GB) is still missing
df = df.dropna(subset=["Storage_GB"])

print("\n=== 3. CLEANED DATAFRAME ===")
print(df)
```
