# Exercise 111: Only the Words

## Objective
Create a function that extracts all words from a string while removing leading and trailing punctuation marks.

---

## Requirements & Rules
1. **Punctuation Removal**: Strip punctuation marks from the **edges** of words.
   * Marks to remove: `,`, `.`, `?`, `-`, `'`, `’`, `!`, `:`, `;`
2. **Preserve Internal Punctuation**: Do **not** remove punctuation marks inside words (e.g., contractions like `don’t`, `isn’t`, or hyphenated words like `state-of-the-art`).
3. **Reusable Design**: Ensure the main execution code runs under `if __name__ == "__main__":` so this script can be safely imported as a module in future exercises (like Exercise 158).

---

## Example
* **Input**: `"Examples of contractions include: don’t, isn’t, and wouldn’t."`
* **Output**: `["Examples", "of", "contractions", "include", "don’t", "isn’t", "and", "wouldn’t"]`


```python
def only_words(text):
    punctuation = ",.?-'’!:;"
    raw_words = text.split()
    cleaned_words = []
    for word in raw_words:
        cleaned = word.strip(punctuation)
        if cleaned:
            cleaned_words.append(cleaned)
    return cleaned_words
while True:
    user_input = input("Enter a sentence: ")
    if user_input.strip() == "":
        print("\nDone!")
        break
    result = only_words(user_input)
    print(f"Extracted Words: {result}\n")
```
