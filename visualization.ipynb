```python
import matplotlib.pyplot as plt

# Define category labels and color list separately
x = ["A", "B", "C"]
colors = ["red", "green", "blue"]

# Initialize a figure with 1 row and 3 columns of subplots
fig, axes = plt.subplots(1, 3, figsize=(10, 4))

# Plot each bar chart and pass the color list
axes[0].bar(x, [1, 2, 3], color=colors)
axes[0].set_title("Bar chart1")

axes[1].bar(x, [5, 3, 2], color=colors)
axes[1].set_title("Bar chart2")

axes[2].bar(x, [4, 7, 1], color=colors)
axes[2].set_title("Bar chart3")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
n = 12

df = pd.DataFrame({
    "month": pd.date_range("2024-01-01", periods=n, freq="MS").strftime("%b"),
    "sales": rng.integers(50, 100, n),
    "cost": rng.integers(20, 60, n),
    "region": rng.choice(["East", "West"], size=n)
})

df["profit"] = df["sales"] - df["cost"]
df
```

```python
fig,ax = plt.subplots()
ax.plot(df["month"], df["sales"], color="blue", linestyle="-", linewidth=2, marker="o", label="sales")
ax.plot(df["month"], df["cost"], color="green", linestyle="--", linewidth=2, marker="s", label="cost")
ax.plot(df["month"], df["profit"], color="orange", linestyle=":", linewidth=2, marker="^", label="Profit")
ax.set_title("Line Chart For the Annual amount of Sales, Cost and Profit")
ax.set_xlabel("Month")
ax.set_ylabel("Amount")
ax.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(12,6))

n = len(df["month"])
x = np.arange(n) # The label locations
width = 0.25 # The width of the bars

sales = df["sales"]
cost = df["cost"]
profit = df["profit"]
months = df["month"]

ax.bar(x - width, sales, width, label="Sales", color="blue")
ax.bar(x, cost, width, label="Cost", color="red")
ax.bar(x + width, profit, width, label="Profit", color="orange")

ax.set_xticks(x)
ax.set_xticklabels(months)
ax.set_title("Sales, Cost, and Profit over Months")
ax.set_xlabel("Month")
ax.set_ylabel("Amount")
ax.legend()

plt.tight_layout()
plt.show()
```
