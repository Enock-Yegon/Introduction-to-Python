```python
import pandas as pd

# Load dataset
df = pd.read_csv('/content/used_cars_data.csv')

# Preview the raw columns and first 3 rows
print("Raw columns:", df.columns.tolist())
df.head(3)
```

```python
# Remove the redundant serial number column
df = df.drop(columns=['S.No.'], errors='ignore')

# Verify column count
print(f"Remaining columns count: {len(df.columns)}")
df.head(3)
```

```python
# Extract brand name (first word of 'Name')
df['Brand'] = df['Name'].str.split().str[0].str.title()

# Compare Name vs Brand
df[['Name', 'Brand']].head()
```

```python
# Extract numbers from text like "26.6 km/kg" or "19.67 kmpl"
df['Mileage_kmpl'] = df['Mileage'].str.extract(r'([\d\.]+)').astype(float)

# Compare old vs cleaned Mileage
df[['Mileage', 'Mileage_kmpl']].head()
```

```python
# Extract numbers from text like "998 CC"
df['Engine_CC'] = df['Engine'].str.extract(r'([\d\.]+)').astype(float)

# Compare old vs cleaned Engine
df[['Engine', 'Engine_CC']].head()
```

```python
# Replace 'null bhp' with None, then extract numeric values
df['Power_bhp'] = (
    df['Power']
    .replace('null bhp', None)
    .str.extract(r'([\d\.]+)')
    .astype(float)
)

# Compare old vs cleaned Power
df[['Power', 'Power_bhp']].head()
```

```python
# Drop original string columns with units
df = df.drop(columns=['Mileage', 'Engine', 'Power'], errors='ignore')

# Save cleaned dataset to a new CSV file
df.to_csv('/content/cleaned_used_cars_data.csv', index=False)

# Display final summary and preview
print("=== CLEANED DATA SUMMARY ===")
print(df.info())

# Display the final cleaned DataFrame output in Colab
df.head(10)
```
