## Intro to Pandas: Data Structures & Basics

### Overview
**Pandas** is an open-source library built on top of NumPy. It introduces two primary data structures that make handling tabular data fast and intuitive:

1. **Series**: A 1D labeled array (like a single column in Excel).
2. **DataFrame**: A 2D labeled data structure with rows and columns (like an entire Excel sheet or SQL table).

---

### Core Data Structures
| Structure | Dimensions | Description |
| :--- | :--- | :--- |
| **Series** | 1D | Single column of homogenous or mixed data with an index. |
| **DataFrame** | 2D | Tabular spreadsheet with row indices and column labels. |

---

### Key Learning Objectives
* Creating DataFrames from dictionaries.
* Inspecting datasets (`head()`, `info()`, `describe()`).
* Selecting specific columns and filtering rows with conditions.

```python
import pandas as pd

# 1. Create a simple dataset using a Python dictionary
data = {
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Enock"],
    "Department": ["HR", "IT", "Data Science", "IT", "Data Science"],
    "Score": [88, 42, 95, 68, 91],
    "Passed": [True, False, True, True, True],
}

# 2. Convert dictionary to a Pandas DataFrame
df = pd.DataFrame(data)

# 3. View the first few rows
print("=== DATAFRAME OVERVIEW ===")
print(df)

# 4. Quick inspection tools
print("\n=== SUMMARY INFO ===")
df.info()

print("\n=== STATISTICAL SUMMARY ===")
print(df.describe())

# 5. Filtering rows (e.g., Score >= 70)
print("\n=== HIGH SCORERS (Score >= 70) ===")
high_scorers = df[df["Score"] >= 70]
print(high_scorers)
```
